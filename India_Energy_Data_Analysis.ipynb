```python
# Year wise averge YoY % change -Plotly lne chart
# How does the year on year percentage changes in capacity vary over the year

import plotly.express as px
# filter capacity data in MW
capacity_data =df[
    (df['Category']== 'Capacity')&
    (df['Unit']== 'MW')
] 
# calculate average YOY % change for each year
yoy_analysis =(
    capacity_data.groupby('Year')['YoY % change']
    .mean()
    .reset_index()
)  
# Line chart
fig = px.line(yoy_analysis,
              x='Year',
              y='YoY % change',
              markers=True,
              title='Year wise Average YoY % change in capacity'
             )
fig.update_layout(
    xaxis_title= 'Year',  
    yaxis_title='Average YoY % change'
)
fig.show()
```

Line plot -Year wise averge YoY % change.

 YoY % change : YoY (Year-over-Year) % Change shows the percentage increase or decrease compared with the previous year. A positive value indicates         growth, while a negative value indicates a decline in capacity.
 
 YoY % Change = ((Current Year Value − Previous Year Value) / Previous Year Value) × 100
 
2020 (~5%) – Capacity showed a small positive average growth.
2021 (~103%) – There was an exceptionally large average YoY growth rate, the highest in the period.
2022 (~5%) – The average growth rate dropped sharply compared with 2021, but remained positive.
2023 (~1–3%) – Growth slowed further and stayed close to zero.
2024 (negative) – The average YoY change became slightly negative, indicating an average decline relative to the previous year.

Insight: The average YoY percentage change in capacity shows a major peak in 2021, indicating exceptionally high average capacity growth during that year. After 2021, the growth rate declined sharply. By 2023, growth was close to zero, while 2024 shows a slight negative average change. Overall, the chart indicates that capacity growth was highly volatile in 2021 and became much slower in the following years.

```python
# Scatter plot : relationship between energy value and YoY % change
```

```python
import matplotlib.pyplot as plt

plot_df = df[
    (df['Value'] >= 0) &
    (df['Value'] <= 5000) &
    (df['YoY % change'] >= -100) &
    (df['YoY % change'] <= 100)
]

plt.figure(figsize=(10,6))

plt.scatter(
    plot_df['Value'],
    plot_df['YoY % change'],
    alpha=0.5
)

plt.xlabel('Value')
plt.ylabel('YoY % change')
plt.title('Value vs YoY % change')
plt.grid(True, alpha=0.3)

plt.show()
```

Scatter plot :

X-axis: Represents the Value of the energy-related observations.
Y-axis: Represents the Year-over-Year (YoY) percentage change.

Insights:
* Lower values show a wider variation in YoY % change, ranging approximately from -100% to +100%.
* Higher values are mostly concentrated around 0% YoY change, indicating relatively smaller percentage fluctuations.
* Positive YoY % values indicate an increase, while negative values indicate a decrease compared with the previous year.
* There is no clear linear relationship between Value and YoY % change.
* Some extreme positive and negative values indicate large year-over-year fluctuations.
